# Spotify Track Popularity - EDA (Lab 1)

```
import os
import pandas as pd, matplotlib.pyplot as plt
if os.path.basename(os.getcwd())=='notebooks': os.chdir('..')
df=pd.read_csv('data/raw/spotify_tracks.csv')
print(df.shape); df.head()
```

```
print(df.dtypes); print(df.isna().sum()[lambda s:s>0])
```

```
print('duplicate track_id rows:', df.track_id.duplicated().sum())
df=df.drop_duplicates('track_id')
df['is_popular']=(df.popularity>=50).astype(int)
print(df.is_popular.value_counts(normalize=True))
```

```
df.popularity.hist(bins=50); plt.axvline(50,color='r'); plt.title('Popularity distribution (red = popular threshold)'); plt.show()
```

```
num=['duration_ms','danceability','energy','loudness','speechiness','acousticness','instrumentalness','liveness','valence','tempo']
df[num+['is_popular']].corr()['is_popular'].drop('is_popular').sort_values().plot.barh(title='Correlation with is_popular'); plt.show()
```

```
df.groupby('explicit').is_popular.mean().plot.bar(title='Popular rate by explicit'); plt.show()
df.groupby('track_genre').is_popular.mean().sort_values().tail(15).plot.barh(title='Top 15 genres by popular rate'); plt.show()
```
